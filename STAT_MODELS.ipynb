```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# Membuat DataFrame secara langsung di dalam sel
dataku = {
    "Partisipan": np.arange(1, 11),
    "Peng Makanan": [50, 40, 60, 55, 45, 65, 70, 75, 80, 90],
    "Peng Transportasi": [20, 25, 30, 35, 40, 50, 55, 60, 65, 70],
    "Berat Badan": [60, 55, 65, 70, 62, 75, 80, 85, 90, 95]
}
df = pd.DataFrame(dataku)

# Mendefinisikan variabel independen (X) dan dependen (y)
X = df[["Peng Makanan", "Peng Transportasi"]]
y = df["Berat Badan"]
X = sm.add_constant(X)

# Membuat model regresi OLS
model = sm.OLS(y, X)
results = model.fit()

# Mengambil koefisien regresi dari hasil fit
intercept = results.params["const"]
coef_makanan = results.params["Peng Makanan"]
coef_transportasi = results.params["Peng Transportasi"]

r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

# Membuat persamaan regresi
persamaan = (
    f"Ŷ = {intercept:.3f}"
    f" + ({coef_makanan:.3f}) X1"
    f" + ({coef_transportasi:.3f}) X2"
)

print("\n")
print("=" * 70)
print(" PERSAMAAN REGRESI")
print("=" * 70)
print(persamaan)
print(f"\nR2 = {r_squared:.4f}")
print(f"Adjusted R2 = {adjusted_r_squared:.4f}")
```

```python
# Menghitung Nilai Prediksi menggunakan variabel X yang sudah ada
df["Prediksi"] = results.predict(X)
print("\n")
print("=" * 70)
print(" DATA AKTUAL & PREDIKSI")
print("=" * 70)
print(
    df[["Partisipan", "Peng Makanan", "Peng Transportasi", "Berat Badan", "Prediksi"]].round(2)
)
```

# Uji Normalitas Residual Regresi

Dalam analisis regresi, asumsi normalitas mengharuskan nilai **residual** (bukan variabel independen atau dependen secara terpisah) berdistribusi normal.

Di bawah ini adalah kode untuk menghitung residual, melakukan uji statistik Jarque-Bera, serta memplot grafik Q-Q Plot dan Histogram.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import statsmodels.api as sm

# Membuat ulang DataFrame dan mendefinisikan model agar variabel 'results' tersedia
dataku = {
    "Partisipan": np.arange(1, 11),
    "Peng Makanan": [50, 40, 60, 55, 45, 65, 70, 75, 80, 90],
    "Peng Transportasi": [20, 25, 30, 35, 40, 50, 55, 60, 65, 70],
    "Berat Badan": [60, 55, 65, 70, 62, 75, 80, 85, 90, 95]
}
df = pd.DataFrame(dataku)

X = df[["Peng Makanan", "Peng Transportasi"]]
y = df["Berat Badan"]
X = sm.add_constant(X)

model = sm.OLS(y, X)
results = model.fit()

# Menghitung Residual
residual = results.resid
df["Residual"] = residual

# Uji Formal Normalitas (Jarque-Bera)
jb_stat, jb_p_value = stats.jarque_bera(residual)

print("=" * 60)
# UJI FORMAL NORMALITAS (JARQUE-BERA)
print(f"Statistik Jarque-Bera : {jb_stat:.4f}")
print(f"p-value               : {jb_p_value:.4f}")
print("-" * 60)
if jb_p_value > 0.05:
    print("Kesimpulan: Residual berdistribusi NORMAL (p-value > 0.05)")
else:
    print("Kesimpulan: Residual TIDAK berdistribusi normal (p-value <= 0.05)")
print("=" * 60)
```

```python
# Visualisasi Deteksi Normalitas
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Grafik 1: Histogram Residual + Kurva Normal
sns_installed = True
try:
    import seaborn as sns
    sns.histplot(residual, kde=True, ax=axes[0], color='skyblue', edgecolor='black')
except ImportError:
    # Fallback jika seaborn tidak terpasang
    axes[0].hist(residual, bins=5, edgecolor='black', alpha=0.7, density=True, color='skyblue')
    # Tambahkan estimasi kurva kepadatan sederhana
    xmin, xmax = axes[0].get_xlim()
    x_axis = np.linspace(xmin, xmax, 100)
    y_axis = stats.norm.pdf(x_axis, np.mean(residual), np.std(residual))
    axes[0].plot(x_axis, y_axis, color='red', linewidth=2)

# Grafik 2: Q-Q Plot (Normal Probability Plot)
sm.qqplot(residual, line='s', ax=axes[1])
axes[1].set_title('Normal Q-Q Plot')
axes[1].grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Memuat data Kota Bandung
df_pe = pd.read_csv("KOTA BANDUNG.csv")

# Mengambil data dari DataFrame (sesuaikan nama DataFrame jika berbeda)
tahun = df_pe["Tahun"]
bm_infrastruktur = df_pe["bm_infrastruktur"]
bm_ekonomi = df_pe["bm_ekonomi"]
nilai_y = df_pe["PE"]  # Variabel dependen Y

# Melakukan regresi linear dengan matriks
X = np.column_stack(
    (bm_infrastruktur, bm_ekonomi)
)  # Matriks X dengan variabel independen
X = np.column_stack(
    (np.ones(len(tahun)), X)
)  # Tambahkan kolom untuk intercept (Beta 0)
beta = (
    np.linalg.inv(X.T @ X) @ X.T @ nilai_y
)  # Hitung parameter beta (koefisien regresi)

print("Koefisien Regresi:")
for i in range(len(beta)):
  print(f"Beta{i}: {beta[i]}")

# Plotting data aktual dan hasil regresi
fig, ax = plt.subplots()
ax.scatter(tahun, nilai_y, color="blue", label="Data Aktual (PE)")
ax.plot(
    tahun, X @ beta, color="red", label="Hasil Regresi"
)  # Garis hasil prediksi regresi
ax.set_xlabel("Tahun")
ax.set_ylabel("PE")
ax.legend()
plt.show()

# INTERPRETASI:
# Persamaan Regresi: PE = 2.946 - 0.008 * bm_infrastruktur + 0.014 * bm_ekonomi
# 1. Konstanta (Beta0 = 2.946):
#    Jika belanja modal infrastruktur dan ekonomi bernilai 0, maka Pertumbuhan
#    Ekonomi (PE) Kota Bandung diproyeksikan sebesar 2.946%
# 2. Belanja Modal Infrastruktur (Beta1 = -0.008):
#    Memiliki hubungan negatif yang sangat kecil terhadap PE. Setiap peningkatan
#    satu unit belanja infrastruktur cenderung menurunkan PE sebesar 0.008%,
#    namun dampak ini secara statistik relatif tidak signifikan.
# 3. Belanja Modal Ekonomi (Beta2 = 0.014):
#    Memiliki hubungan positif terhadap PE. Setiap peningkatan satu unit belanja
#    ekonomi diproyeksikan meningkatkan PE sebesar 0.014%.
# Visualisasi menunjukkan bahwa model regresi (garis merah) mampu menangkap
# fluktuasi umum dari data aktual PE (titik biru), meskipun terdapat deviasi
# pada tahun-tahun tertentu (seperti penurunan tajam di tahun 2020 akibat pandemi).
# ==============================================================================
```
