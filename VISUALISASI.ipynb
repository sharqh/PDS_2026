# Visualisasi Grafik Berdasarkan Dataset

#DATASET 1

```python
from matplotlib import pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Muat dataset (1)
df_penjualan = pd.read_csv("https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv")
df_penjualan

# Membersihkan nama kolom terlebih dahulu
df_penjualan.columns = df_penjualan.columns.str.strip()

# Menambah Kolom baru "total"
df_penjualan["total"] = df_penjualan[['jan','feb','mar','apr']].sum(axis=1)
df_penjualan
```

```python
# Membuat Bar Chart Per Orang
plt.figure(figsize=(8, 4))
df_sorted = df_penjualan.sort_values("total", ascending=False)
sns.barplot(data=df_sorted, x="nama", y="total", palette="viridis")
plt.title("Total Penjualan per Orang")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Grafik ini menunjukkan total penjualan tiap karyawan selama empat bulan (Januari–April),
# diurutkan dari yang tertinggi ke terendah. Terlihat bahwa Indara menempati posisi puncak
# dengan total penjualan paling tinggi, disusul oleh Chandra dan Dudi di bawahnya.
#Ketiganya bisa dibilang sebagai top performer karena angkanya cukup jauh di atas rekan-rekan yang lain.
# Selanjutnya ada Budi dan Joni di posisi menengah, sementara Eko, Gina, Herman, Andi, dan Fahri berada
# di kelompok bawah dengan total penjualan yang lebih kecil. Fahri menjadi yang paling rendah,
# dengan selisih hampir empat kali lipat dibanding Indara.
```

```python
import seaborn as sns
import numpy as np
import pandas as pd

# Muat dataset (2)
df_fiskal = pd.read_csv("https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv")
df_fiskal

# Point Plot (Tren Rata-rata per Tahun)
plt.figure(figsize=(8, 4))
sns.pointplot(data=df_fiskal, x="Tahun", y="KEMISKINAN",
              color="green", marker="D")
plt.title("Tren Rata-rata Kemiskinan per Tahun")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Distribusi & Hubungan Sederhana
plt.figure(figsize=(8, 4))
sns.histplot(data=df_fiskal, x="KEMISKINAN", bins=30, kde=True, color="navy")
plt.title("Distribusi Tingkat Kemiskinan di Seluruh Daerah")
plt.xlabel("Tingkat Kemiskinan (%)")
plt.ylabel("Frekuensi (Jumlah Daerah)")
plt.tight_layout()
plt.show()
```
